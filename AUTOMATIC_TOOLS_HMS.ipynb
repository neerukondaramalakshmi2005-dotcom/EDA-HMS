| Tool                | Main Purpose                  | What it gives                                                    |
| ------------------- | ----------------------------- | ---------------------------------------------------------------- |
| **D-Tale**          | Interactive data exploration  | Data table, statistics, charts, missing values                   |
| **YData Profiling** | Automatic data profiling      | Complete EDA report, correlations, missing values, distributions |
| **AutoViz**         | Automatic visualization       | Automatically generates useful graphs                            |
| **Sweetviz**        | Dataset analysis & comparison | Visual EDA report, distributions, correlations, comparisons      |


```python
import pandas as pd

A = pd.read_csv("/content/appointments.csv")
B = pd.read_csv("/content/billing.csv")
D = pd.read_csv("/content/doctors.csv")
P = pd.read_csv("/content/patients.csv")
T = pd.read_csv("/content/treatments.csv")
print(A)
print(B)
print(D)
print(P)
print(T)
```

```python
! pip install dtale
```

```python
import dtale
import pandas as pd

# Open and display interactive D-Tale sheets for the dataframes
dtale.show(A)  # Appointments
dtale.show(B)  # Billing
dtale.show(D)  # Doctors
dtale.show(P)  # Patients
dtale.show(T)  # Treatments
```

```python
!pip install ydata-profiling
```

```python
from ydata_profiling import ProfileReport

ProfileReport(A, title="Appointments Report")
ProfileReport(B, title="Billing Report")
ProfileReport(D, title="Doctors Report")
ProfileReport(P, title="Patients Report")
ProfileReport(T, title="Treatments Report")
```

```python
! pip install autoviz
```

```python
# Note: The wordcloud warning below is normal for non-text/date columns and can be ignored.
!python -m textblob.download_corpora

from autoviz.AutoViz_Class import AutoViz_Class

AV = AutoViz_Class()

AV.AutoViz("/content/appointments.csv", sep=",", depVar="status", dfte=A)
AV.AutoViz("/content/billing.csv", sep=",", depVar="amount", dfte=B)
AV.AutoViz("/content/doctors.csv", sep=",", depVar="specialization", dfte=D)
AV.AutoViz("/content/patients.csv", sep=",", depVar="date_of_birth", dfte=P)
AV.AutoViz("/content/treatments.csv", sep=",", depVar="cost", dfte=T)
```
