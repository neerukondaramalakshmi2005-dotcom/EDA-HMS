# Interpolate means numerical values to find it
# the datasets

```python
datasets = {
    "Appointments": A,
    "Billing": B,
    "Doctors": D,
    "Patients": P,
    "Treatments": T
}

for name, df in datasets.items():
    # Fill numeric columns by interpolation
    numeric_cols = df.select_dtypes(include="number").columns
    df[numeric_cols] = df[numeric_cols].interpolate()

    # Fill categorical columns with their most frequent value (mode)
    categorical_cols = df.select_dtypes(include=["object", "category"]).columns
    for col in categorical_cols:
        if df[col].isna().any() and not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])
```

# # IDENTIFY THE OUTLIERS

```python
# IDENTIFY IQR OUTLIER
#============Appointment============
# Re-calculating bounds specifically for appointment_time
temp_time = pd.to_timedelta(A['appointment_time'])
q1_time = temp_time.quantile(0.25)
q3_time = temp_time.quantile(0.75)
iqr_time = q3_time - q1_time
lower_time = q1_time - 1.5 * iqr_time
upper_time = q3_time + 1.5 * iqr_time

iqr_outliers_A = A[(temp_time < lower_time) | (temp_time > upper_time)]
print("Appointments IQR Outliers:", len(iqr_outliers_A))

#=============Billing============
# Re-calculating bounds specifically for billing amount
q1_bill = B['amount'].quantile(0.25)
q3_bill = B['amount'].quantile(0.75)
iqr_bill = q3_bill - q1_bill
lower_bill = q1_bill - 1.5 * iqr_bill
upper_bill = q3_bill + 1.5 * iqr_bill

iqr_outliers_B = B[(B['amount'] < lower_bill) | (B['amount'] > upper_bill)]
print("Billing IQR Outliers:", len(iqr_outliers_B))

#===============Patient================
# Using date_of_birth (converted to datetime) for identifying age/DOB outliers
temp_dob = pd.to_datetime(P['date_of_birth'])
q1_dob = temp_dob.quantile(0.25)
q3_dob = temp_dob.quantile(0.75)
iqr_dob = q3_dob - q1_dob
lower_dob = q1_dob - 1.5 * iqr_dob
upper_dob = q3_dob + 1.5 * iqr_dob

iqr_outliers_P = P[(temp_dob < lower_dob) | (temp_dob > upper_dob)]
print("Patient DOB IQR Outliers:", len(iqr_outliers_P))

#=================Doctor================
# Detecting outliers for Doctor's years_experience
q1_doc = D['years_experience'].quantile(0.25)
q3_doc = D['years_experience'].quantile(0.75)
iqr_doc = q3_doc - q1_doc
lower_doc = q1_doc - 1.5 * iqr_doc
upper_doc = q3_doc + 1.5 * iqr_doc

iqr_outliers_D = D[(D['years_experience'] < lower_doc) | (D['years_experience'] > upper_doc)]
print("Doctor Experience IQR Outliers:", len(iqr_outliers_D))

# ===================treatment===================
# Detecting outliers for Treatment cost
q1_treat = T['cost'].quantile(0.25)
q3_treat = T['cost'].quantile(0.75)
iqr_treat = q3_treat - q1_treat
lower_treat = q1_treat - 1.5 * iqr_treat
upper_treat = q3_treat + 1.5 * iqr_treat

iqr_outliers_T = T[(T['cost'] < lower_treat) | (T['cost'] > upper_treat)]
print("Treatment Cost IQR Outliers:", len(iqr_outliers_T))
```

# Z-score method (it is define as the lower case to upper case)

```python
import pandas as pd
from scipy.stats import zscore
```

```python
# Convert the string appointment_time to total seconds to make it numeric
numeric_time = pd.to_timedelta(A['appointment_time']).dt.total_seconds()

# Calculate the z-scores and add it as a new column
A['appointment_time_zscore'] = zscore(numeric_time)

# Find outliers where the absolute z-score is greater than 3
z_outliers = A[A['appointment_time_zscore'].abs() > 3]
print("z-score outliers:", len(z_outliers))
```

```python
from scipy.stats import zscore
import pandas as pd

# Calculate temporary Patient age based on date_of_birth to make it numeric
P['age'] = 2024 - pd.to_datetime(P['date_of_birth']).dt.year

datasets = {
    "Billing": (B, "amount"),
    "Doctors": (D, "years_experience"),
    "Patients": (P, "age"),
    "Treatments": (T, "cost")
}

for name, (df, column) in datasets.items():
    # Skip columns that don't exist or aren't numeric
    if column not in df.columns or not pd.api.types.is_numeric_dtype(df[column]):
        print(f"{name}: check the column name; '{column}' must be numeric.")
        continue

    # Calculate z-scores, ignoring missing values
    z_scores = zscore(df[column], nan_policy="omit")
    df[f"{column}_zscore"] = z_scores

    # Find rows more than 3 standard deviations from the mean
    z_outliers = df[df[f"{column}_zscore"].abs() > 3]
    print(f"{name} z-score outliers:", len(z_outliers))
```

# 3. z-score Trimming  ( analysis of dataset )



```python
from scipy.stats import zscore
import pandas as pd

# Initialize the trimmed_datasets dictionary to avoid the NameError
trimmed_datasets = {}

# Define the target datasets and their numeric columns for z-score trimming
targets = {
    "Billing": (B, "amount"),
    "Doctors": (D, "years_experience"),
    "Patients": (P, "age"),
    "Treatments": (T, "cost")
}

# Loop through each to calculate z-scores and trim outliers (|z| <= 3)
for name, (df, column) in targets.items():
    scores = pd.Series(zscore(df[column], nan_policy="omit"), index=df.index)
    trimmed_df = df[scores.notna() & (scores.abs() <= 3)]
    trimmed_datasets[name] = trimmed_df

    print(f"--- {name} ({column}) ---")
    print("Rows before trimming:", len(df))
    print("Rows after trimming:", len(trimmed_df))
    print("Rows removed:", len(df) - len(trimmed_df))
    print()

# Now this line will execute without NameError
P_trimmed = trimmed_datasets["Patients"]


#==============appointment==================
from scipy.stats import zscore
import pandas as pd

numeric_time = pd.to_timedelta(A["appointment_time"]).dt.total_seconds()
z_scores = zscore(numeric_time, nan_policy="omit")

keep_rows = pd.Series(z_scores, index=A.index).notna() & (
    pd.Series(z_scores, index=A.index).abs() <= 3
)

A_trimmed = A.loc[keep_rows].copy()

print("Appointments removed:", len(A) - len(A_trimmed))
print("Appointments remaining:", len(A_trimmed))
```

# Winsorization (it is used for exit value)

```python
# 5. winsorization  ( it is used for extreme values / outliers )
from scipy.stats.mstats import winsorize
import pandas as pd

# =============Appointment===================
# Since reason_for_visit is non-numeric, we winsorize a numeric representation of appointment_time instead
A['appointment_time_seconds'] = pd.to_timedelta(A['appointment_time']).dt.total_seconds()
A['appointment_time_winsorize'] = winsorize(A['appointment_time_seconds'], limits=[0, 0.3])
print("--- Appointments (Winsorized Time in Seconds) ---")
print(A[['appointment_time', 'appointment_time_winsorize']].head())
print()
print(A[['appointment_time', 'appointment_time_winsorize']].tail(11))
print()


# =================Billing===================
# Winsorizing the numeric 'amount' column
B['amount_winsorize'] = winsorize(B['amount'], limits=[0, 0.3])
print("--- Billing (Winsorized Amount) ---")
print(B[['amount', 'amount_winsorize']].head())
print(B[['amount', 'amount_winsorize']].tail(8))

# ==============Patient======================
# Patients has no 'cost' column. We will winsorize the numeric 'age' column instead
P['age_winsorize'] = winsorize(P['age'], limits=[0, 0.3])
print("--- Patients (Winsorized Age) ---")
print(P[['age', 'age_winsorize']].head())
print(P[['age', 'age_winsorize']].tail(9))


# ================Doctor===================
D['years_experience_winsorize'] = winsorize(D['years_experience'], limits=[0, 0.3])
print("--- doctor(Winsorized years_experience) ---")
print(D[['years_experience', 'years_experience_winsorize']].head())
print(D[['years_experience', 'years_experience_winsorize']].tail(11))

#=============Treatment====================
T['cost_winsorize'] = winsorize(T['cost'], limits=[0, 0.3])
print("--- Treatment(Winsorized cost) ---")
print(T[['cost', 'cost_winsorize']].head())
print(T[['cost', 'cost_winsorize']].tail(10))

```
